```python
df['Time'] = pd.to_datetime(df['Time'])
df.sort_values(by="Time", inplace=True)
df.reset_index(drop=True, inplace=True)
df.head()
```

```python
euro_to_drachma = df[['Time','Greek drachma']].copy()
euro_to_drachma['Greek drachma'].value_counts()
```

```python
euro_to_drachma.info()
```

```python
euro_to_drachma = euro_to_drachma[euro_to_drachma['Greek drachma'] != '-']
euro_to_drachma['Greek drachma'] = euro_to_drachma['Greek drachma'].astype(float)
euro_to_drachma.info()
```

### Visualizations

```python
fig = px.line(euro_to_drachma, x="Time", y="Greek drachma")
fig.update_xaxes(range=["1999-01-01", "2000-12-31"])
fig.show()
```

```python
import plotly.graph_objs as go

fig = go.Figure()

# Original values
fig.add_trace(
    go.Scatter(
        x=euro_to_drachma['Time'],
        y=euro_to_drachma['Greek drachma'],
        mode='lines',
        name="Original"
    )
)

# Rolling averages
for window in [7, 30, 50, 100, 365]:
    fig.add_trace(
        go.Scatter(
            x=euro_to_drachma['Time'],
            y=euro_to_drachma['Greek drachma'].rolling(window).mean(),
            mode='lines',
            name=f"Rolling {window}"
        )
    )

# Layout customization
fig.update_layout(
    title="Euro to Greek Drachma - Rolling Window Analysis",
    xaxis_title="Time",
    yaxis_title="Greek Drachma",
    template="plotly_white",
    height=600,
    width=1000
)

fig.update_xaxes(range=["1999-01-01", "2000-12-31"])

fig.show()
```
